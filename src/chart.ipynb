```python
"""
chart.py
pyecharts 交互图

功能：
    折线图：城北 COD 月均值 + 国标阈值线
"""
from pyecharts.charts import Line
from pyecharts import options as opts




# 折线图	pyecharts 城北COD月均值图表+阈值线

# 筛选城北 + COD，按月份分组算均值，保留 2 位，重置索引。

def plot_cod_line(df_merge2, output='output/test_cod_bar.html'):
    py_dff = df_merge2[(df_merge2['site_name']=='城北污水厂') & (df_merge2['indictor_name']=='COD')].groupby('month')['value'].mean().round(2).reset_index()
    print(py_dff)
    # 创建折线图对象
    line = Line()
    # X 轴月份（转字符串），Y 轴均值
    line.add_xaxis(py_dff['month'].astype(str).to_list()).add_yaxis('城北 COD 月均值',py_dff['value'].to_list())
    # 在 y=60 处画一条红色虚线，标"cod阈值"
    line.set_series_opts(
        markline_opts = opts.MarkLineOpts(
            data=[opts.MarkLineItem(y=50, name='cod阈值')],
            linestyle_opts=opts.LineStyleOpts(color='red', type_='dashed')
        )
    )

    # 全局配置
    line.set_global_opts(
        title_opts=opts.TitleOpts(title="城北 COD 月均值",pos_left="center",pos_bottom="1%"),
        legend_opts=opts.LegendOpts(is_show=True,pos_top="1%"),
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        xaxis_opts=opts.AxisOpts(name='月份'),
        yaxis_opts=opts.AxisOpts(name='cod/mg/L')
    )
    # 生成 HTML 文件
  
    line.render(output)



if __name__ == '__main__':
    from clean_analyze import load_data, iqr_1, add_time_columns

    df_merge2 = load_data()
    df_merge2= df_merge2.groupby(['site_id', 'indictor_id'], group_keys=False).apply(iqr_1)
    df_merge2= add_time_columns(df_merge2)

    plot_cod_line(df_merge2)
    print('图表生成完成，见 output/')


```
